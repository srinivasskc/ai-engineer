Output Parsers

Output Parsers in langchain help convert raw LLM responses into structured format like JSON, CSV, Pydantic Models, and more.
They ensure consistency, validation and ease of use in applications


Output Parsers can be used with models, where structured outputs and unstructured outputs are there as well.

Types of Output Parsers
1. StringOutputParser
2. JsonOutputParser
3. StructuredOutputParser
4. PydanticOutputParser
5. CSVOutputParser
6. ListOutputParser


String Output Parser:

It is simplest output parser in langchain, It is used to parse the output of LLM and return it as plain string

Open Source Models does not produce structured outputs. Example: HuggingFace -> TinyLlama models
